## E91 Demonstration

The below cell runs a demonstration of the E91 protocol. Note that due to classical observations it is not possible for $S$ to converge to the theoretical limit of $2\sqrt{2}$.

```python
import random

# Configuration
n = 20000  # Number of entangled pairs
angles_alice = [0, 45, 90]         # Alice's measurement settings (in degrees)
angles_bob = [22.5, 67.5, 90]      # Bob's measurement settings (in degrees)

# Bell test settings
chsh_settings = {
    (0, 22.5): 'A1B1',
    (0, 67.5): 'A1B2',
    (45, 22.5): 'A2B1',
    (45, 67.5): 'A2B2'
}

# Set this to True to simulate eavesdropping
eve_active = False  # <- Toggle this

# Generate entangled bits (1 or 0, perfect correlation)
def generate_entangled_pairs(n):
    return [random.randint(0, 1) for _ in range(n)]

# Randomly select measurement bases for Alice and Bob
def choose_measurement_basis(n, possible_angles):
    return [random.choice(possible_angles) for _ in range(n)]

# Simulate quantum measurement
def simulate_measurement(entangled_bit, angle_a, angle_b, eve=False):
    if eve:
        # Eve's interference adds noise
        if random.random() < 0.5:
            entangled_bit = 1 - entangled_bit
    if angle_a == angle_b:
        return entangled_bit, entangled_bit
    else:
        correlation = 0.75  # simulate quantum-like imperfect correlation
        if random.random() < correlation:
            return entangled_bit, entangled_bit
        else:
            return entangled_bit, 1 - entangled_bit

# Calculate expectation value
def expectation(values):
    return sum(values) / len(values) if values else 0

# Main simulation
entangled_bits = generate_entangled_pairs(n)
alice_bases = choose_measurement_basis(n, angles_alice)
bob_bases = choose_measurement_basis(n, angles_bob)

alice_results = []
bob_results = []
chsh_counts = {'A1B1': [], 'A1B2': [], 'A2B1': [], 'A2B2': []}

# Run measurements
for i in range(n):
    a_angle = alice_bases[i]
    b_angle = bob_bases[i]
    a_bit, b_bit = simulate_measurement(entangled_bits[i], a_angle, b_angle, eve=eve_active)
    alice_results.append(a_bit)
    bob_results.append(b_bit)

    if (a_angle, b_angle) in chsh_settings:
        label = chsh_settings[(a_angle, b_angle)]
        a_val = 1 if a_bit == 0 else -1
        b_val = 1 if b_bit == 0 else -1
        chsh_counts[label].append(a_val * b_val)

# Calculate CHSH inequality
E_A1B1 = expectation(chsh_counts['A1B1'])
E_A1B2 = expectation(chsh_counts['A1B2'])
E_A2B1 = expectation(chsh_counts['A2B1'])
E_A2B2 = expectation(chsh_counts['A2B2'])
S = abs(E_A1B1 + E_A1B2 + E_A2B1 - E_A2B2)

# Output results
print("E91 Quantum Key Distribution with Bell Test")
print("Eavesdropping active:", eve_active)
print(f"Number of entangled pairs: {n}")
print(f"CHSH S value: {S:.3f}")
print("Expectations:")
print(f"  E(A1, B1): {E_A1B1:.3f}")
print(f"  E(A1, B2): {E_A1B2:.3f}")
print(f"  E(A2, B1): {E_A2B1:.3f}")
print(f"  E(A2, B2): {E_A2B2:.3f}")
print()
if S > 2:
    print("✔ Bell inequality violated — Quantum correlations intact.")
else:
    print("✘ No Bell violation — Possible eavesdropping or classical behavior.")
```
