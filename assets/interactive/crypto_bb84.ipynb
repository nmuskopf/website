## BB84 Demonstration

The below cell runs a demonstration of the BB84 protocol.

```python
import random

def generate_bits(n):
    return [random.randint(0, 1) for _ in range(n)]

def generate_bases(n):
    return [random.choice(['+', 'x']) for _ in range(n)]

def encode_qubits(bits, bases):
    return list(zip(bits, bases))

def measure_qubits(qubits, bases):
    results = []
    for (bit, qubit_basis), measurement_basis in zip(qubits, bases):
        if qubit_basis == measurement_basis:
            results.append(bit)
        else:
            results.append(random.randint(0,1))
    return results

def sift_keys(sender_bases, receiver_bases, receiver_bits):
    key = []
    sifted_indices = []
    for i in range(len(sender_bases)):
        if sender_bases[i] == receiver_bases[i]:
            key.append(receiver_bits[i])
            sifted_indices.append(i)
    return key, sifted_indices

def simulate_eavesdropping(qubits):
    eve_bases = generate_bases(len(qubits))
    intercepted = []
    for (bit, basis), eve_basis in zip(qubits, eve_bases):
        if basis == eve_basis:
            measured_bit = bit
        else:
            measured_bit = random.randint(0,1)
        intercepted.append((measured_bit, eve_basis))
    return intercepted

def estimate_error_rate(alice_key, bob_key, sample_size=10):
    if len(alice_key) < sample_size:
        return None, None
    sample_indices = random.sample(range(len(alice_key)), sample_size)
    alice_sample = [alice_key[i] for i in sample_indices]
    bob_sample = [bob_key[i] for i in sample_indices]
    errors = sum(a != b for a, b in zip(alice_sample, bob_sample))
    error_rate = errors / sample_size
    return error_rate, sample_indices

n = 100

alice_bits = generate_bits(n)
alice_bases = generate_bases(n)
qubits = encode_qubits(alice_bits, alice_bases)

qubits_after_eve = simulate_eavesdropping(qubits)
print(qubits)
print(qubits_after_eve)

bob_bases = generate_bases(n)
bob_results = measure_qubits(qubits_after_eve, bob_bases)

sifted_key_alice, sift_indices = sift_keys(alice_bases, bob_bases, bob_results)
sifted_key_bob = [bob_results[i] for i in sift_indices]

error_rate, sample_indices = estimate_error_rate(sifted_key_alice, sifted_key_bob)

print("Alice's original bits (first 20):", alice_bits[:20])
print("Alice's bases (first 20):        ", alice_bases[:20])
print("Bob's bases (first 20):          ", bob_bases[:20])
print("Bob's results (first 20):        ", bob_results[:20])

print("Sifted key length:", len(sifted_key_alice))
print("Estimated error rate (from sample):", error_rate)
if error_rate is not None and error_rate > 0.2:
    print("High error rate - possible eavesdropping detected!")
else:
    print("Key likely secure.")
print("Sample of Alice's sifted key:", sifted_key_alice[:20])
print("Sample of Bob's sifted key:  ", sifted_key_bob[:20])
```
